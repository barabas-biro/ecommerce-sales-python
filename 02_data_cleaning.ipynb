```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df['Unit Price'] = df['Unit Price'].astype(float)
```

```python
df['Discount'] = df['Discount'].astype(float)
```

```python
df.info()
```

```python
df = df.rename(columns={'Order ID':'order_id','Order Date':'order_date','Customer Name':'customer_name','Region':'region','City':'city','Category':'category','Sub-Category':'sub_category','Product Name':'product_name','Quantity':'quantity','Unit Price':'unit_price','Discount':'discount','Sales':'sales','Profit':'profit','Payment Mode':'payment_mode'})
```

```python
df.columns
```

```python
df = df.assign(year = lambda x : x.order_date.dt.year,
               month = lambda x : x.order_date.dt.month,
               year_month = lambda x : x.order_date.dt.to_period('M'))
```
